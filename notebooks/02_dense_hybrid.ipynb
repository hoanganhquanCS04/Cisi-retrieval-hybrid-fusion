```python
import sys
import json
import faiss
import numpy as np
from pathlib import Path

# Set base project path
project_root = Path('..').resolve()
sys.path.append(str(project_root))

from retrieval.dense_retriever import DenseRetriever
from retrieval.hybrid_pipeline import run_hybrid_pipeline
```

```python
# 1. Đọc dữ liệu Corpus và Query do Member 1 chuẩn bị
corpus_path = project_root / 'data' / 'corpus.json'
queries_path = project_root / 'data' / 'queries.json'

with open(corpus_path, 'r', encoding='utf-8') as f:
    corpus = json.load(f)
    
with open(queries_path, 'r', encoding='utf-8') as f:
    queries = json.load(f)

print(f"Loaded {len(corpus)} documents and {len(queries)} queries.")
```

```python
'''
# 2. Khởi tạo Dense Retriever và Build FAISS Index
retriever = DenseRetriever(model_path_or_name='all-MiniLM-L6-v2', device='c')
embeddings = retriever.build_index(corpus)

output_dir = Path('/kaggle/working/faiss')
output_dir.mkdir(parents=True, exist_ok=True)

retriever.save_index(
    output_dir / 'faiss_index.bin',
    output_dir / 'doc_ids.json'
)

'''
```

```python
retriever = DenseRetriever(model_path_or_name='all-MiniLM-L6-v2', device='cpu')

# Thay vì build_index, ta TẢI INDEX ĐÃ LƯU TỪ TRƯỚC
retriever.load_index(
    faiss_path = project_root / 'data' / 'faiss' / 'faiss_index.bin', 
    doc_ids_path = project_root / 'data' / 'faiss' / 'doc_ids.json'
)
print("Đã tải thành công FAISS Index!")
```

```python
# 3. Tính toán top 100 kết quả cho mỗi query bằng Dense Retrieval
dense_top100 = {}
for q in queries:
    qid = str(q['query_id'])
    text = q['text']
    results = retriever.retrieve(text, top_k=100)
    dense_top100[qid] = [{"doc_id": doc_id, "score": float(score)} for doc_id, score in results]

with open(project_root / 'data' / 'dense_top100.json', 'w', encoding='utf-8') as f:
    json.dump(dense_top100, f, indent=2)

print("Hoàn thành xuất dense_top100.json")
```

```python
# 4. Chạy Hybrid Pipeline (RRF)
print("Đang áp dụng Reciprocal Rank Fusion...")
run_hybrid_pipeline(
    bm25_top100_path = project_root / 'data' / 'bm25_top100.json',
    dense_top100_path = project_root / 'data' / 'dense_top100.json',
    output_path = project_root / 'data' / 'hybrid_top100.json',
    k = 60
)
print("Quy trình hoàn tất! Sẵn sàng bàn giao cho M3.")
```
